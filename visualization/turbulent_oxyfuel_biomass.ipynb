# turbulent-oxyfuel-biomass

Plots the y=0 slice (`writeSurface`, `slice_y`) and the walnut-shell cloud (`writeCloud`)
written by `flameletCloudFoam` into `turbulent-oxyfuel-biomass/postProcessing/`.
Run the case first (see `../turbulent-oxyfuel-biomass/README.md`), or point `CASE` at existing output.

The slice is contoured on its own mesh triangulation, so the burner block appears as a gap
rather than being filled by interpolation.

```python
import os
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np

import oxysim_viz as ov

CASE = Path(os.environ.get("OXYSIM_TURBULENT_CASE", "../turbulent-oxyfuel-biomass"))
TIME = None  # None = latest available; otherwise the nearest available time is used
PLANE_TOL = 0.03  # m -- particles within this |y| of the slice are drawn

ov.set_style()
```

```python
mesh, t = ov.load_slice(CASE, "writeSurface", "slice_y.vtp", TIME)
cloud, t_cloud = ov.load_cloud(CASE, "writeCloud" ,"walnutCloud", t)
print(f"slice t = {t}, cloud t = {t_cloud}, {cloud.n_points} particles")
print("slice fields:", list(mesh.point_data.keys()))
```

## Temperature with particles

```python
fig, ax = plt.subplots(figsize=(6, 9), dpi=200)
ov.plot_slice(ax, mesh, "T", cloud=cloud, cloud_color="T", plane_tol=PLANE_TOL,
              cmap="inferno", particle_size_by="d")
ax.set_title(f"t = {t} s")
plt.show()
```

## Gas-phase fields

```python
fields = {"T": "inferno", "OH": "plasma", "Z": "viridis", "|U|": "magma"}
fig, axs = plt.subplots(1, len(fields), figsize=(4.5 * len(fields), 8), dpi=200, gridspec_kw={"wspace": 0.9})
for ax, (f, cmap) in zip(axs, fields.items()):
    ov.plot_slice(ax, mesh, f, cmap=cmap)
    ax.set_title(f)
plt.show()
```
